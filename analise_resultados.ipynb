```python
import pandas as pd
import matplotlib.pyplot as plt

lat = pd.read_csv('latencia.csv')
vaz = pd.read_csv('vazao.csv')
lat, vaz
```

```python
# Gráfico Latência
plt.figure()
plt.plot(lat['host_pair'], lat['latencia_ms'])
plt.xlabel('Par de Hosts')
plt.ylabel('Latência (ms)')
plt.title('Latência por Par de Hosts')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Gráfico Vazão
plt.figure()
plt.plot(vaz['host_pair'], vaz['vazao_Mbps'])
plt.xlabel('Par de Hosts')
plt.ylabel('Vazão (Mbps)')
plt.title('Vazão por Par de Hosts')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Comparação Latência x Vazão
plt.figure()
plt.plot(lat['host_pair'], lat['latencia_ms'], label='Latência (ms)')
plt.plot(vaz['host_pair'], vaz['vazao_Mbps'], label='Vazão (Mbps)')
plt.legend()
plt.title('Comparativo Latência x Vazão')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
