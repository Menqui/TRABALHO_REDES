```python
import pandas as pd
import matplotlib.pyplot as plt

lat = pd.read_csv('latencia.csv')
vaz = pd.read_csv('vazao.csv')
trafego_switch = pd.read_csv('trafego_switch.csv')

# Criar coluna 'host_pair'
lat['host_pair'] = lat['Origem'] + '-' + lat['Destino']
vaz['host_pair'] = vaz['Origem'] + '-' + vaz['Destino']

# Converter valores numéricos (N/A vira NaN)
lat['Latência Média (ms)'] = pd.to_numeric(lat['Latência Média (ms)'], errors='coerce')
vaz['Vazão TCP (Mbps)'] = pd.to_numeric(vaz['Vazão TCP (Mbps)'], errors='coerce')

lat.head(), vaz.head()
```

```python
#  Gráfico de Latência
plt.figure()
plt.plot(lat['host_pair'], lat['Latência Média (ms)'])
plt.xlabel('Par de Hosts')
plt.ylabel('Latência (ms)')
plt.title('Latência por Par de Hosts')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
#  Gráfico de Vazão
plt.figure()
plt.plot(vaz['host_pair'], vaz['Vazão TCP (Mbps)'])
plt.xlabel('Par de Hosts')
plt.ylabel('Vazão TCP (Mbps)')
plt.title('Vazão TCP por Par de Hosts')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
#  Comparação Latência x Vazão
plt.figure()
plt.plot(lat['host_pair'], lat['Latência Média (ms)'], label='Latência (ms)')
plt.plot(vaz['host_pair'], vaz['Vazão TCP (Mbps)'], label='Vazão TCP (Mbps)')
plt.legend()
plt.title('Comparativo Latência x Vazão')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```
